Customer satisfaction is influenced by multiple factors—response time, resolution quality, communication tone, product usability, and post-purchase support—each interacting in ways that are difficult to model with linear techniques. By contrast, the ANN architecture is designed to process large volumes of data, identify subtle correlations, and adaptively learn from diverse inputs. Data normalization and encoding ensure that the model can interpret both numerical and categorical variables effectively. Once prepared, the dataset is divided into training, validation, and testing subsets to evaluate the model’s performance across different scenarios.

The ANN architecture employed in this project consists of multiple hidden layers, each designed to capture increasingly abstract representations of customer data. Activation functions such as ReLU (Rectified Linear Unit) are used to introduce non-linearity, while optimization techniques like Adam are applied to minimize prediction error. Regularization methods, including dropout, are incorporated to prevent overfitting and enhance generalization. The model is trained iteratively, with performance metrics such as Mean Squared Error (MSE) and accuracy monitored to fine-tune hyperparameters.

Results from the predictive model demonstrate significant improvements over baseline statistical approaches. The ANN consistently achieves higher accuracy in forecasting CSAT scores, particularly in cases where satisfaction is influenced by multiple overlapping factors. These insights allow organizations to prioritize interventions, allocate resources more effectively, and design targeted strategies to improve customer experience.

Beyond predictive accuracy, the project emphasizes the practical utility of deep learning in customer experience management. By pinpointing satisfaction drivers, businesses can implement data-driven policies such as enhancing support training, optimizing response workflows, or redesigning product features. The ability to forecast dissatisfaction before it escalates empowers organizations to take preventive measures, thereby reducing churn and strengthening long-term customer relationships.

In conclusion, this project illustrates the transformative potential of deep learning in e-commerce. The ANN-based predictive model not only advances the measurement of customer satisfaction but also provides a strategic tool for continuous improvement. By integrating advanced analytics into customer service operations, businesses can move beyond reactive approaches and embrace proactive, intelligent customer experience management. This data-driven methodology positions organizations to thrive in a competitive environment, ensuring sustained growth through enhanced customer loyalty and satisfaction.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#Calculating missing value percentage fo every variable
missing_percent = (data.isnull().sum() / len(data)) * 100
print(missing_percent.sort_values(ascending=False))
````

````python
for column in data.columns:
    # Columns having more than 70% missing values
    if missing_percent[column] > 70:
        data.drop(columns=[column], inplace=True)

    # Columns having moderate missing values
    elif 10 < missing_percent[column] <= 70:
        if data[column].dtype == 'object':
            data[column] = data[column].fillna("Not Provided")
        else:
            data[column] = data[column].fillna(data[column].median())

    # Columns with very few missing values
    elif missing_percent[column] > 0:
        if data[column].dtype == 'object':
            data[column] = data[column].fillna(data[column].mode()[0])
        else:
            data[column] = data[column].fillna(data[column].mean())
````

#### Chart - 5

````python
# Chart - 5 visualization code
import pandas as pd
import plotly.express as px

try:
    # Convert to datetime
    data['Issue_reported at'] = pd.to_datetime(
        data['Issue_reported at'],
        dayfirst=True,
        format='mixed',
        errors='coerce'
    )

    data['issue_responded'] = pd.to_datetime(
        data['issue_responded'],
        dayfirst=True,
        format='mixed',
        errors='coerce'
    )

    # Calculate Resolution Time
    data['Resolution Time'] = data['issue_responded'] - data['Issue_reported at']

    # Convert to minutes
    data['Resolution Time (minutes)'] = data['Resolution Time'].dt.total_seconds() / 60

    # Remove invalid or negative values
    data = data[data['Resolution Time (minutes)'] >= 0]

    # Create bins
    bins = [0, 10, 30, 60, 120, 240, 480, data['Resolution Time (minutes)'].max()]
    labels = ['<10 min', '10-30 min', '30-60 min', '60-120 min', '2-4 hours', '4-8 hours', '>8 hours']

    data['Resolution Time Category'] = pd.cut(
        data['Resolution Time (minutes)'],
        bins=bins,
        labels=labels,
        right=False
    )

    # Average CSAT by resolution category
    avg_csat_by_resolution = (
        data.groupby('Resolution Time Category', observed=False)['CSAT Score']
        .mean()
        .reset_index()
    )

    # Plot using Plotly
    fig = px.bar(
        avg_csat_by_resolution,
        x='CSAT Score',
        y='Resolution Time Category',
        orientation='h',
        title="Average CSAT Score by Resolution Time Category",
        labels={
            'CSAT Score': 'Average CSAT Score',
            'Resolution Time Category': 'Resolution Time Category'
        },
        color='CSAT Score',
        color_continuous_scale='Blues'
    )

    fig.show()

except KeyError as e:
    print(f"Column missing in dataset: {e}")

except ValueError as e:
    print(f"Value error occurred: {e}")

except Exception as e:
    print(f"Unexpected error occurred: {e}")
````

This chart shows how CSAT scores vary with the time taken to respond to customer issues. A bar chart makes it easy to compare average CSAT scores across these time buckets.

This visualization analyzes the impact of issue response time on customer satisfaction (CSAT) scores. It highlights that quicker the issue is responded to greater is the customer satisfaction level.

````python
# Chart - 10 visualization code
import plotly.express as px

# Average CSAT per manager
avg_csat_manager = (
    data.groupby('Manager')['CSAT Score']
        .mean()
        .sort_values(ascending=False)
)

# Top 10 managers
top_managers = avg_csat_manager.head(10).index

# Count number of tasks handled by these managers
manager_task_count = (
    data[data['Manager'].isin(top_managers)]
        .groupby('Manager')
        .size()
        .rename('Task Count')
)

# Combine both metrics
manager_summary = (
    avg_csat_manager.loc[top_managers]
        .to_frame('Avg CSAT')
        .join(manager_task_count)
        .reset_index()
)

# Scatter plot
fig = px.scatter(
    manager_summary,
    x='Task Count',
    y='Avg CSAT',
    text='Manager',
    title='Top Managers: CSAT Score vs Tasks Handled',
    labels={
        'Task Count': 'Number of Tasks Assigned',
        'Avg CSAT': 'Average CSAT Score'
    },
    size='Task Count'
)

fig.update_traces(textposition='top center')

fig.show()
````

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation
data = pd.read_csv('/content/eCommerce_Customer_support_data.csv')
missing_percent = (data.isnull().sum() / len(data)) * 100
print("Missing values in the data")
print(missing_percent.sort_values(ascending=False))
for column in data.columns:
    # Columns having more than 70% missing values
    if missing_percent[column] > 70:
        data.drop(columns=[column], inplace=True)

    # Columns having moderate missing values
    elif 10 < missing_percent[column] <= 70:
        if data[column].dtype == 'object':
            data[column] = data[column].fillna("Not Provided")
        else:
            data[column] = data[column].fillna(data[column].median())

    # Columns with very few missing values
    elif missing_percent[column] > 0:
        if data[column].dtype == 'object':
            data[column] = data[column].fillna(data[column].mode()[0])
        else:
            data[column] = data[column].fillna(data[column].mean())
print(f"Total missing values in the data after data preprocessing: {data.isnull().sum().sum()}")
````

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Target variable
target = data['CSAT Score']

# Calculate quartiles
Q1 = target.quantile(0.25)
Q3 = target.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detect outliers
outliers = data[(target < lower_bound) | (target > upper_bound)]

print("Number of outliers:", len(outliers))
print(outliers)

# Winsorization for handling these outliers
data['CSAT Score'] = data['CSAT Score'].clip(lower_bound, upper_bound)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Outliers are values in the dataset which differ largely from the mean are average values of the data. This may introduce noise in the data which affects the training of the model. Therefore handling outliers is an important step in data preprocessing.

IQR technique - This method finds values that are far away from the middle 50% of the data. Any data less than 25th percentile or greater than 75th percentile of the data is considered as an outlier.
Winsorization - These outliers are capped to the boundry values(either 25% or 75%) for data handling. This method is best suited for numerical data.

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
import pandas as pd
import re
import nltk
import contractions

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required resources
nltk.download('stopwords')
nltk.download('wordnet')

# Load dataset
# data = pd.read_csv("Customer_support_data.csv") # Removed redundant data load

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def normalize_text(text):
    if pd.isna(text):
        return text

    # 1. Expand contractions (can't → cannot)
    text = contractions.fix(text)

    # 2. Lowercase
    text = text.lower()

    # 3. Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # 4. Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # 5. Remove words containing digits
    text = re.sub(r'\b\w*\d\w*\b', '', text)

    # 6. Remove stopwords
    words = text.split()
    words = [w for w in words if w not in stop_words]

    # 7. Lemmatization
    words = [lemmatizer.lemmatize(w) for w in words]

    # 8. Remove extra whitespace
    text = " ".join(words).strip()

    return text

# Apply normalization only to Customer Remarks
data['Customer_Remarks_Normalized'] = data['Customer Remarks'].apply(normalize_text)
pd.DataFrame({
    "Original": data['Customer Remarks'].head(5),
    "Normalized": data['Customer_Remarks_Normalized'].head(5)
})
````

````python
# ML Model - 2 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# ================================
# Improved ANN Model for CSAT
# ================================

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam
import numpy as np

# ------------------------------
# Custom Focal Loss
# ------------------------------
def focal_loss(gamma=2., alpha=0.25):

    def loss(y_true, y_pred):

        y_true = tf.one_hot(tf.cast(y_true, tf.int32), depth=5)

        epsilon = 1e-7
        y_pred = tf.clip_by_value(y_pred, epsilon, 1. - epsilon)

        cross_entropy = -y_true * tf.math.log(y_pred)

        weight = alpha * tf.pow((1 - y_pred), gamma)

        loss = weight * cross_entropy

        return tf.reduce_mean(tf.reduce_sum(loss, axis=1))

    return loss


# ================================
# Build Tuned ANN
# ================================
from tensorflow.keras.layers import Input

model = Sequential([

Input(shape=(X_train_balanced_scaled.shape[1],)),

Dense(128,
      activation='relu',
      kernel_regularizer=l2(0.001)),

BatchNormalization(),
Dropout(0.4),

Dense(64,
      activation='relu',
      kernel_regularizer=l2(0.001)),

BatchNormalization(),
Dropout(0.35),

Dense(32,
      activation='relu'),

Dropout(0.25),

Dense(16,
      activation='relu'),

Dense(5, activation='softmax')

])


# ================================
# Compile Model
# ================================

model.compile(

optimizer=Adam(learning_rate=0.0005),

loss=focal_loss(gamma=2, alpha=0.25),

metrics=['accuracy']

)


# ================================
# Early Stopping
# ================================

early_stop = EarlyStopping(

monitor='val_loss',

patience=12,

restore_best_weights=True

)


# ================================
# Train Model
# ================================

history = model.fit(

X_train_balanced_scaled,
y_train_class,

validation_split=0.2,

epochs=150,

batch_size=64,

callbacks=[early_stop],

class_weight=class_weights,

verbose=1

)


# ================================
# Predictions
# ================================

y_pred_prob = model.predict(X_test_scaled)

y_pred_class = np.argmax(y_pred_prob, axis=1)

y_pred = y_pred_class + 1

y_test_original_scale = y_test_class + 1
````

````python
# hyper parameter tuning for ANN model
import keras_tuner as kt
from tensorflow import keras
from tensorflow.keras import layers

def build_ann(hp):

    # Input layer
    inputs = keras.Input(shape=(X_train_balanced_scaled.shape[1],))

    x = layers.Dense(
        hp.Int("dense1", 128, 512, step=64),
        activation="relu"
    )(inputs)

    x = layers.BatchNormalization()(x)

    x = layers.Dropout(
        hp.Float("dropout1", 0.2, 0.5, step=0.05)
    )(x)

    x = layers.Dense(
        hp.Int("dense2", 64, 256, step=32),
        activation="relu"
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.Dropout(
        hp.Float("dropout2", 0.2, 0.5, step=0.05)
    )(x)

    x = layers.Dense(
        hp.Int("dense3", 32, 128, step=16),
        activation="relu"
    )(x)

    outputs = layers.Dense(
        len(np.unique(y_train_class_ann)),
        activation="softmax"
    )(x)
    # Model tuning
    model = keras.Model(inputs, outputs)

    model.compile(
        optimizer=keras.optimizers.Adam(
            hp.Choice("learning_rate", [1e-2,1e-3,5e-4,1e-4])
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
    tuner = kt.RandomSearch(
    build_ann,
    objective="val_accuracy",
    max_trials=2,
    executions_per_trial=1,
    directory="csat_tuning",
    project_name="ann_tuning"
)

tuner.search(
    X_train_balanced_scaled,
    y_train_class_ann,
    epochs=100,
    validation_split=0.2,
    batch_size=64
)
# Selecting best model among all trials
best_ann = tuner.get_best_models(num_models=1)[0]
````

````python
# hyperparameter tuning for XGBoost
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier

xgb = XGBClassifier(
    objective="multi:softprob",
    num_class=len(np.unique(y_train_class_xgb)),
    eval_metric="mlogloss"
)

param_grid = {

    "n_estimators":[200,300,400,500],
    "max_depth":[4,5,6,7,8],
    "learning_rate":[0.01,0.03,0.05,0.1],
    "subsample":[0.7,0.8,0.9,1],
    "colsample_bytree":[0.7,0.8,0.9,1],
    "gamma":[0,0.1,0.2],
    "reg_lambda":[1,1.5,2]
}

search = RandomizedSearchCV(
    xgb,
    param_grid,
    n_iter=25,
    scoring="accuracy",
    cv=3,
    verbose=1,
    n_jobs=-1
)

search.fit(X_train_balanced_scaled, y_train_class_xgb)

best_xgb = search.best_estimator_
````
